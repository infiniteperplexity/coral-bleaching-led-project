```python
synthetic = df.copy()
synthetic['dhw'] = synthetic.apply(lambda x: scale_dhw(int(x['year']), int(x['month']), int(x['day'])), axis=1)
#synthetic['dhw'] = 0.0
synthetic.head()
```

```python
synthetic.to_csv('data/synthetic_all.csv', index=False)
#synthetic.to_csv('data/synthetic_all_zero.csv', index=False)
```
